```python
%sql
-- Databricks notebook source
-- ==============================================================================
-- PHASE 10: Unity Catalog Registration (Mapping ADLS External Locations to SQL)
-- ==============================================================================

-- 1. Create the Catalog with an explicit Managed Location to bypass the root URL error
CREATE CATALOG IF NOT EXISTS portfolio_12026
MANAGED LOCATION 'abfss://bronze@datalakeseniorproj012026.dfs.core.windows.net/managed_catalog/';

-- 2. Set the newly created catalog as the active context
USE CATALOG portfolio_12026;

-- 3. Create logical schemas mapped to the Medallion architecture
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- 4. Register Bronze Layer External Tables
CREATE TABLE IF NOT EXISTS bronze.lineitem 
USING DELTA 
LOCATION 'abfss://bronze@datalakeseniorproj012026.dfs.core.windows.net/lineitem/';

CREATE TABLE IF NOT EXISTS bronze.orders 
USING DELTA 
LOCATION 'abfss://bronze@datalakeseniorproj012026.dfs.core.windows.net/orders/';

-- 5. Register Silver Layer External Tables
CREATE TABLE IF NOT EXISTS silver.lineitem 
USING DELTA 
LOCATION 'abfss://silver@datalakeseniorproj012026.dfs.core.windows.net/lineitem/';

CREATE TABLE IF NOT EXISTS silver.orders 
USING DELTA 
LOCATION 'abfss://silver@datalakeseniorproj012026.dfs.core.windows.net/orders/';

-- 6. Register Gold Layer External Tables
CREATE TABLE IF NOT EXISTS gold.sales_summary 
USING DELTA 
LOCATION 'abfss://gold@datalakeseniorproj012026.dfs.core.windows.net/sales_summary/';

-- 7. Verify the final business aggregate is queryable via standard SQL
SELECT * FROM gold.sales_summary LIMIT 10;
```
